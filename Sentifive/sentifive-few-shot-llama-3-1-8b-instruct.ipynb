```python
import os
import random
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

from kaggle_secrets import UserSecretsClient

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig
)

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    classification_report,
    confusion_matrix
)
```

```python
SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

DATA_PATH = "/kaggle/input/datasets/amermahbub01/sentifive/SentiFive.csv"

MODEL_NAME = "meta-llama/Llama-3.1-8B-Instruct"

N_SAMPLES = 3000
RANDOM_SEED = 42
MAX_NEW_TOKENS = 20

LABELS = [
    "STRONGLY NEGATIVE",
    "SLIGHTLY NEGATIVE",
    "NEUTRAL",
    "SLIGHTLY POSITIVE",
    "STRONGLY POSITIVE"
]

print("Model:", MODEL_NAME)
print("Evaluation samples:", N_SAMPLES)
print("Random seed:", RANDOM_SEED)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
df = pd.read_csv(DATA_PATH)

print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())

display(df.head())
```

```python
label_mapping = {
    "নিশ্চিত নেতিবাচক": "STRONGLY NEGATIVE",
    "কিছুটা নেতিবাচক": "SLIGHTLY NEGATIVE",
    "নিরপেক্ষ": "NEUTRAL",
    "কিছুটা ইতিবাচক": "SLIGHTLY POSITIVE",
    "নিশ্চিত ইতিবাচক": "STRONGLY POSITIVE"
}

df["label"] = df["title"].map(label_mapping)

print("Label distribution:")
print(df["label"].value_counts(dropna=False))
```

```python
unknown_labels = df.loc[
    df["label"].isna(),
    "title"
].unique()

if len(unknown_labels) > 0:
    print("Unmapped labels found:")
    print(unknown_labels)
else:
    print("All labels mapped successfully.")
```

```python
eval_df = (
    df.sample(
        n=N_SAMPLES,
        random_state=RANDOM_SEED
    )
    .reset_index(drop=True)
)

print("Evaluation dataset shape:", eval_df.shape)

print("\nEvaluation label distribution:")
print(eval_df["label"].value_counts())

display(eval_df.head())
```

```python
demo_pool = df.drop(
    index=eval_df.index,
    errors="ignore"
).copy()
```

```python
eval_indices = df.sample(
    n=N_SAMPLES,
    random_state=RANDOM_SEED
).index

eval_df = (
    df.loc[eval_indices]
    .reset_index(drop=True)
)

demo_pool = df.drop(
    index=eval_indices
).copy()

print("Evaluation dataset shape:", eval_df.shape)
print("Demonstration pool shape:", demo_pool.shape)

print("\nEvaluation label distribution:")
print(eval_df["label"].value_counts())

print("\nDemonstration pool label distribution:")
print(demo_pool["label"].value_counts())
```

```python
user_secrets = UserSecretsClient()

HF_TOKEN = user_secrets.get_secret("gptoss")

print("Hugging Face token loaded successfully.")
```

```python
tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME,
    token=HF_TOKEN
)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

print("Tokenizer loaded successfully.")
```

```python
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True
)

print("4-bit quantization configured.")
```

```python
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    token=HF_TOKEN,
    quantization_config=bnb_config,
    device_map="auto",
    torch_dtype=torch.float16
)

model.eval()

print("Llama 3.1 8B Instruct loaded successfully.")
```

```python
def few_shot_messages(text):

    system_prompt = """You are a Bengali sentiment classification system.

Classify the sentiment of the given Bengali text into exactly ONE of these five labels:

STRONGLY NEGATIVE
SLIGHTLY NEGATIVE
NEUTRAL
SLIGHTLY POSITIVE
STRONGLY POSITIVE

Use the examples below to understand how Bengali sentiment is classified.

Return ONLY the final sentiment label.
Do not provide explanations.
Do not provide any additional text.
"""

    messages = [
        {
            "role": "system",
            "content": system_prompt
        }
    ]

    for example in few_shot_examples:

        messages.append({
            "role": "user",
            "content": (
                "Bengali text:\n\n"
                f"{example['text']}\n\n"
                "Sentiment label:"
            )
        })

        messages.append({
            "role": "assistant",
            "content": example["label"]
        })

    messages.append({
        "role": "user",
        "content": (
            "Bengali text:\n\n"
            f"{text}\n\n"
            "Sentiment label:"
        )
    })

    return messages
```

```python
def parse_sentiment(output):

    if output is None:
        return "UNKNOWN"

    text = output.strip().upper()

    if text in LABELS:
        return text

    cleaned = (
        text
        .replace(":", "")
        .replace(".", "")
        .replace("-", " ")
        .strip()
    )

    if cleaned in LABELS:
        return cleaned

    for label in LABELS:
        if label in text:
            return label

    aliases = {
        "STRONGLY NEGATIVE": [
            "STRONGLY_NEGATIVE",
            "STRONG NEGATIVE"
        ],
        "SLIGHTLY NEGATIVE": [
            "SLIGHTLY_NEGATIVE",
            "MILD NEGATIVE"
        ],
        "NEUTRAL": [
            "NEUTRAL"
        ],
        "SLIGHTLY POSITIVE": [
            "SLIGHTLY_POSITIVE",
            "MILD POSITIVE"
        ],
        "STRONGLY POSITIVE": [
            "STRONGLY_POSITIVE",
            "STRONG POSITIVE"
        ]
    }

    for label, variants in aliases.items():

        for variant in variants:

            if variant in text:
                return label

    return "UNKNOWN"
```

```python
def generate_response(
    messages,
    max_new_tokens=20
):

    inputs = tokenizer.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        return_tensors="pt",
        return_dict=True
    )

    inputs = {
        key: value.to(model.device)
        for key, value in inputs.items()
    }

    with torch.no_grad():

        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )

    input_length = inputs["input_ids"].shape[-1]

    generated_tokens = outputs[0][input_length:]

    response = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    )

    return response.strip()
```

```python
sample_text = eval_df.iloc[0]["data"]

messages = few_shot_messages(
    sample_text
)

response = generate_response(
    messages,
    max_new_tokens=MAX_NEW_TOKENS
)

prediction = parse_sentiment(response)

print("Bengali text:")
print(sample_text)

print("\nRaw model output:")
print(response)

print("\nParsed prediction:")
print(prediction)

print("\nActual label:")
print(eval_df.iloc[0]["label"])
```

```python
few_shot_predictions = []
few_shot_raw_outputs = []

print("Starting Few-Shot evaluation...\n")

for i, text in enumerate(
    eval_df["data"]
):

    messages = few_shot_messages(text)

    output = generate_response(
        messages,
        max_new_tokens=MAX_NEW_TOKENS
    )

    prediction = parse_sentiment(output)

    few_shot_predictions.append(prediction)
    few_shot_raw_outputs.append(output)

    if (i + 1) % 100 == 0:

        print(
            f"Processed "
            f"{i + 1}/{len(eval_df)} samples"
        )

print("\nFew-Shot evaluation completed.")
```

```python
few_shot_results_df = eval_df.copy()

few_shot_results_df["prediction"] = (
    few_shot_predictions
)

few_shot_results_df["raw_output"] = (
    few_shot_raw_outputs
)

print(
    "Results shape:",
    few_shot_results_df.shape
)

display(
    few_shot_results_df[
        [
            "data",
            "label",
            "prediction",
            "raw_output"
        ]
    ].head(10)
)
```

```python
few_shot_y_true = few_shot_results_df["label"]
few_shot_y_pred = few_shot_results_df["prediction"]

few_shot_accuracy = accuracy_score(
    few_shot_y_true,
    few_shot_y_pred
)

(
    few_shot_precision,
    few_shot_recall,
    few_shot_f1,
    _
) = precision_recall_fscore_support(
    few_shot_y_true,
    few_shot_y_pred,
    labels=LABELS,
    average="macro",
    zero_division=0
)

print("=" * 50)
print("FEW-SHOT RESULTS")
print("=" * 50)

print(f"Accuracy : {few_shot_accuracy:.4f}")
print(f"Precision: {few_shot_precision:.4f}")
print(f"Recall   : {few_shot_recall:.4f}")
print(f"Macro F1 : {few_shot_f1:.4f}")
```

```python
print(
    classification_report(
        few_shot_y_true,
        few_shot_y_pred,
        labels=LABELS,
        zero_division=0,
        digits=4
    )
)
```

```python
(
    class_precision,
    class_recall,
    class_f1,
    class_support
) = precision_recall_fscore_support(
    few_shot_y_true,
    few_shot_y_pred,
    labels=LABELS,
    zero_division=0
)

few_shot_class_metrics_df = pd.DataFrame({
    "Class": LABELS,
    "Precision": class_precision,
    "Recall": class_recall,
    "F1-Score": class_f1,
    "Support": class_support
})

display(
    few_shot_class_metrics_df
)
```
